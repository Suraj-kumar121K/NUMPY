# STUDENT PERFORMANCE ANALYTICS SYSTEM

# CREATE DATASET

## Import Numpy

```python
import numpy as np
```

```python
np.random.seed(42)
marks = np.random.randint(0,101,(50, 5))
print(marks.shape)
```

## SUBJECT NAMES

```python
subjects = np.array(["Math", "Science", "English", "Hindi", "Social Science"])
```

## BASIC ANALYSIS

1. Total marks per student

```python
total = marks.sum(axis=1)
```

2. Average marks per student

```python
avg = marks.mean(axis=1)
```

3. Max marks per student

```python
max_marks = marks.max(axis=1)
```

### FIND TOPPER

```python
topper_index = np.argmax(total)
print("Topper Student Index:", topper_index)
print("Topper Total Marks:", total[topper_index])
```

## FAIL ANALYSIS

```python
fail_students = np.sum(avg < 40)

print("Failed Students:", fail_students)
```

# SUBJECT ANALYSIS
### Average of each subject

```python
subject_avg = marks.mean(axis=0)

print(subject_avg)
```

HARD SUBJECT FINDING

```python
hard_subject_index = np.argmin(subject_avg)

print("Hardest Subject:", subjects[hard_subject_index])
```

TOP 5 STUDENTS

```python
top5_indices = np.argsort(total)[-5:]

print("Top 5 Students:", top5_indices)
print("Their Scores:", total[top5_indices])
```

PERFORMANCE LABELING

```python
performance = np.where(avg >= 75, "A",
               np.where(avg >= 50, "B",
               np.where(avg >= 40, "C", "F")))

print(performance[:10])
```

SUBJECT WISE BEST STUDENT

```python
best_students = np.argmax(marks, axis=0)

print("Best in each subject:", best_students)
```
