```python
import numpy as np
```

```python
image = np.array([
    [10, 10, 10, 10, 10],
    [10, 50, 50, 50, 10],
    [10, 50,100, 50, 10],
    [10, 50, 50, 50, 10],
    [10, 10, 10, 10, 10]
])
```

```python
kernel_x = np.array([
    [-1, 0, 1],
    [-2, 0, 2],
    [-1, 0, 1]
])
```

```python
kernel_y = np.array([
    [-1, -2, -1],
    [ 0,  0,  0],
    [ 1,  2,  1]
])
```

```python
def convolve(image, kernel):
    h, w = image.shape
    kh, kw = kernel.shape
    
    pad = kh // 2
    padded = np.pad(image, pad, mode='constant')
    
    output = np.zeros((h, w))
    
    for i in range(h):
        for j in range(w):
            region = padded[i:i+kh, j:j+kw]
            output[i, j] = np.sum(region * kernel)
    
    return output
```

```python
edge_x = convolve(image, kernel_x)
edge_y = convolve(image, kernel_y)

# Combine both directions
edges = np.sqrt(edge_x**2 + edge_y**2)

print("Edges:\n", edges)
```

```python
edges = (edges / np.max(edges)) * 255
edges = edges.astype(np.uint8)

print("Final Edge Image:\n", edges)
```
